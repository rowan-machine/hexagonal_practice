This notebook demonstrates the simple database connection utility for analysts.

**Note**: Prefer using SDK classes (ClaimsAnalyst, PoliciesAnalyst) over direct database access.

````python
from src.utils.db_connection import connect_to_warehouse

# Simple connection - automatically finds warehouse.db
db = connect_to_warehouse()

print("Connected to warehouse database")
print(f"Database path: {db.db_path}")
````

## Query Data

Query data directly (only if SDK doesn't provide what you need).


````python
# Query claims data
results = db.query("SELECT * FROM claims_silver LIMIT 5")

print(f"Retrieved {len(results)} records")
if results:
    print("\nFirst record:")
    for key, value in results[0].items():
        print(f"  {key}: {value}")
````
