# Combined Claims and Policies Analysis

This notebook demonstrates how to use the StopLossAnalyst SDK to perform combined analysis.

## Overview

**Important**: Analysts should ONLY use SDK methods. All business logic is encapsulated.

The StopLossAnalyst provides high-level methods that:
- Load data from database automatically
- Calculate coverage utilization
- Generate summaries
- Handle all business logic internally

````python
from decimal import Decimal
from src.sdk import StopLossAnalyst

# Initialize the combined analyst
# SDK handles all database connections internally
analyst = StopLossAnalyst(
    db_path="../warehouse.db",
    approval_threshold=Decimal("100000.00")
)

print("StopLossAnalyst SDK initialized")
print("Ready to analyze claims and policies")
````

## Calculate Coverage Utilization

Get coverage utilization for all policies. SDK loads data and performs all calculations.


````python
# Calculate utilization - SDK loads data and does all calculations
utilization = analyst.get_coverage_utilization()

print(f"Coverage utilization for {len(utilization)} policies:\n")
for util in utilization[:5]:  # Show first 5
    print(f"Policy: {util['policy_id']}")
    print(f"  Employer: {util['employer_id']}")
    print(f"  Stop Loss Limit: ${util['stop_loss_limit']:,.2f}")
    print(f"  Total Claims: ${util['total_claims']:,.2f}")
    print(f"  Utilization: {util['utilization_percent']:.2f}%")
    print(f"  Remaining Coverage: ${util['remaining_coverage']:,.2f}")
    print()
````

## Get Claims by Policy Summary

Get summary of claims grouped by policy. SDK handles all data loading and aggregation.


````python
# Get summary - SDK loads data and performs analysis
summary = analyst.get_claims_by_policy_summary()

print(f"Summary for {len(summary)} policies\n")

for policy_id, data in list(summary.items())[:3]:  # Show first 3
    print(f"Policy: {policy_id}")
    if data['policy']:
        policy = data['policy']
        print(f"  Employer: {policy.get('employer_id', 'N/A')}")
        print(f"  Coverage Limit: ${policy.get('stop_loss_limit', 0):,.2f}")
    print(f"  Claim Count: {data['claim_count']}")
    print(f"  Total Claims: ${data['total_claim_amount']:,.2f}")
    print()
````

## High Utilization Policies

Get policies with high utilization. SDK handles the filtering logic.


````python
# Get high utilization policies - SDK does the filtering
high_utilization = analyst.get_high_utilization_policies(threshold_percent=50.0)

print(f"Policies with >50% utilization: {len(high_utilization)}\n")

for util in high_utilization:
    print(f"{util['policy_id']} ({util['employer_id']}): {util['utilization_percent']:.2f}% utilized")
    print(f"  Claims: ${util['total_claims']:,.2f} / Limit: ${util['stop_loss_limit']:,.2f}")
    print()
````
